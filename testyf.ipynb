```python
#!/usr/bin/env python3
"""
测试yfinance批量下载性能对比
对比download方法和ticker.history方法的性能差异
"""

import asyncio
import time
import pandas as pd
from data.sources.yfinance_client import YFinanceClient

async def test_download_performance():
    """测试yfinance download方法的性能"""
    # 创建配置（空配置即可，因为yfinance不需要特殊配置）
    config = {}
    client = YFinanceClient(config)
    
    # 测试股票列表（使用正确的股票代码格式）
    symbols = ["US.AAPL", "US.MSFT", "US.GOOGL", "US.AMZN", "US.TSLA", "US.META", "US.NVDA", "US.NFLX", "US.AMD", "US.INTC"]
    
    print("=== yfinance批量下载性能测试 ===")
    print(f"测试股票数量: {len(symbols)}")
    print(f"股票列表: {symbols}")
    
    # 测试download方法
    print("\n1. 使用download方法（多线程批量下载）:")
    start_time = time.time()
    
    try:
        results = await client.get_batch_data(
            symbols=symbols,
            start_date="2024-01-01",
            end_date="2024-01-31",
            period="daily"
        )
        
        download_time = time.time() - start_time
        print(f"   下载时间: {download_time:.2f}秒")
        print(f"   成功获取: {len(results)}个股票的数据")
        
        # 显示每个股票的数据量
        for symbol, df in results.items():
            print(f"     {symbol}: {len(df)}行数据")
            
    except Exception as e:
        print(f"   download方法失败: {e}")
    
    # 测试逐个下载方法（模拟原来的方式）
    print("\n2. 模拟ticker.history逐个下载方法:")
    start_time = time.time()
    
    try:
        results = {}
        for symbol in symbols:
            try:
                df = await client.get_historical_data(
                    symbol=symbol,
                    start_date="2024-01-01",
                    end_date="2024-01-31",
                    period="daily"
                )
                results[symbol] = df
            except Exception as e:
                print(f"   获取{symbol}失败: {e}")
        
        individual_time = time.time() - start_time
        print(f"   下载时间: {individual_time:.2f}秒")
        print(f"   成功获取: {len(results)}个股票的数据")
        
        # 显示每个股票的数据量
        for symbol, df in results.items():
            print(f"     {symbol}: {len(df)}行数据")
            
        # 性能对比
        if download_time > 0 and individual_time > 0:
            speedup = individual_time / download_time
            print(f"\n性能对比:")
            print(f"   download方法速度是个别下载的 {speedup:.1f} 倍")
            
    except Exception as e:
        print(f"   逐个下载方法失败: {e}")

async def test_large_batch():
    """测试更大批量的性能"""
    # 创建配置（空配置即可，因为yfinance不需要特殊配置）
    config = {}
    client = YFinanceClient(config)
    
    # 更大的股票列表（使用正确的股票代码格式）
    symbols = [
        "US.AAPL", "US.MSFT", "US.GOOGL", "US.AMZN", "US.TSLA", "US.META", "US.NVDA", "US.NFLX", "US.AMD", "US.INTC",
        "US.SPY", "US.QQQ", "US.IWM", "US.DIA", "US.VOO", "US.IVV", "US.VTI", "US.GLD", "US.SLV", "US.USO"
    ]
    
    print("\n=== 大批量性能测试 ===")
    print(f"测试股票数量: {len(symbols)}")
    
    # 测试download方法
    start_time = time.time()
    
    try:
        results = client.get_batch_data(
            symbols=symbols,
            start_date="2024-01-01",
            end_date="2024-01-31",
            period="daily"
        )
        
        download_time = time.time() - start_time
        print(f"download方法 - 下载时间: {download_time:.2f}秒")
        print(f"download方法 - 成功获取: {len(results)}个股票的数据")
        
    except Exception as e:
        print(f"download方法失败: {e}")

test_large_batch()
    
print("\n测试完成！")
```
